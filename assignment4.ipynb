```python
# Task 3: Handling Null Values
# Create a new DataFrame with some null values and fill them
# Your code here:
data = {
    "Student": ["John", "Emma", "Michael", "Sophia", "Daniel"],
    "Marks": [85, np.nan, 90, 78, np.nan],
    "Subject": ["Math", "Science", np.nan, "English", "Math"],
    "Age": [20, 21, np.nan, 19, 22]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)

# fill the missing values
df['Marks'] = df["Marks"].fillna(df['Marks'].mean())
df['Subject'] = df['Subject'].fillna("Not assigned")
df['Age'] = df['Age'].fillna(df['Age'].mean())

print(f"\nDataframe after filling missing values")
print(df)
```

```python
# Task 5: Concatenation and Merging
# Concatenate two DataFrames vertically and merge them horizontally

df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10], 'D': [11, 12]})

# Your code here:
vertically_concat = pd.concat([df1,df2], axis=0, ignore_index = True)

horizontal_merge = pd.concat([df1, df3], axis= 1)

print("Vertical Concatenation:")
print(vertically_concat)

print("\nHorizontal Merging:")
print(horizontal_merge)
```

```python
# Task 9: Iterators and Generators
# Create an iterator for the first 5 even numbers.
# Then, create a generator for the same.
even_numbers = [2,4,6,8,10]
nums_iterator = iter(even_numbers)

for num in nums_iterator:
  print(num)

# Your code here:
def even_generator():
    for num in range(2, 11, 2):
        yield num

print("\nUsing Generator:")
for num in even_generator():
    print(num)
```

```python
# Task 10: Map, Reduce, and Filter
# Use map to square all numbers in a list.
# Use reduce to find the product of all numbers in a list.
# Use filter to get only even numbers from a list.

numbers = [1, 2, 3, 4, 5]

from functools import reduce

numbers = [1, 2, 3, 4, 5]

# 1. Use map() to square all numbers
squared_numbers = list(map(lambda x: x ** 2, numbers))

# 2. Use reduce() to find the product of all numbers
product = reduce(lambda x, y: x * y, numbers)

# 3. Use filter() to get only even numbers
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

# Display results
print("Squared numbers:", squared_numbers)
print("Product of numbers:", product)
print("Even numbers:", even_numbers)
```

```python
# Task 11: Object-Oriented Programming - Creating a Class
# Create a class called 'Rectangle' with attributes 'length' and 'width'.
# Include methods to calculate area and perimeter.
class rectangle:
  def __init__(self, length, width):
     self.length = length
     self.width = width

# Your code here:
  def area(self):
    return self.length * self.width

  def perimeter(self):
    return 2* (self.length + self.width)

result = rectangle(5,7)
print(result.area())
print(result.perimeter())
```

```python

# Task 12: Pandas Data Analysis
# Using the following DataFrame, perform these tasks:
# 1. Find the average salary by department
# 2. Get the names of employees with salary > 60000
# 3. Add a new column 'Bonus' which is 10% of salary

df_employees = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]
})

avg_salary = df_employees.groupby('Department')['Salary'].mean()

print("Average Salary by Department:")
print(avg_salary)


# 2. Get names of employees with salary > 60000
high_salary = df_employees[df_employees['Salary'] > 60000]['Name']

print("\nEmployees earning more than 60000:")
print(high_salary)

# 3. Add a new column Bonus (10% of salary)
df_employees['Bonus'] = df_employees['Salary'] * 0.10

print("\nUpdated Employee DataFrame:")
print(df_employees)
```
